```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("/content/drive/MyDrive/opencv/bird.jpg")

# Check if the image was loaded successfully
if img is None:
    print("Error: Image not loaded. Please check the file path.")
else:
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    alpha = 1.5   # Contrast (1.0 = original)
    beta = 50     # Brightness (0 = original)
    bright_contrast = cv2.convertScaleAbs(img, alpha=alpha, beta=beta)

    # Gaussian Blur
    gaussian = cv2.GaussianBlur(img, (5,5), 0)

    # Median Filter
    median = cv2.medianBlur(img, 5)

    # Show results
    titles = ["Original", "Grayscale", "Brightness+Contrast", "Gaussian Blur", "Median Filter"]
    images = [img, gray, bright_contrast, gaussian, median]

    plt.figure(figsize=(12,6))
    for i in range(5):
        plt.subplot(1,5,i+1)
        if len(images[i].shape) == 2:
            plt.imshow(images[i], cmap="gray")
        else:
            plt.imshow(cv2.cvtColor(images[i], cv2.COLOR_BGR2RGB))
        plt.title(titles[i])
        plt.axis("off")
    plt.show()
```
